```python
import json
import random
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

images_path = Path('dataset/images')
data = json.loads(Path('dataset/gt_public.json').read_text())
image_ids = list(data.keys())

randomly_selected_id = random.sample(image_ids, 15)

fig, axes = plt.subplots(3, 5, figsize=(20, 12))

for ax, image_id in zip(axes.flat, randomly_selected_id):
    image = plt.imread(images_path / f'{image_id}.png')
    ax.imshow(image, cmap='gray')
    ax.set_title(image_id)
    ax.axis('off')

    for x, y, w, h in data[image_id]:
        ax.add_patch(Rectangle((x, y), w, h, fill=False, edgecolor='red', linewidth=1))

plt.show()
```
